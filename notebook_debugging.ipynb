```python
import pandas as pd

from autoembed.src.domain.dataset_preprocessor import DatasetPreprocessor
from autoembed.src.infrastructure.model.embedding_model_keras_adapter import KerasAutoencoder
```

```python
dab_data = pd.read_csv("/Users/pa/Desktop/workspace/autoembed/data/dab/dab-last-versions.csv")
```

```python
dataset_preprocessor = DatasetPreprocessor(
    categorical_columns_names=["category","make", "model", "energy", "gearbox"],
    text_column_name="version"
)

dataset_preprocessor.fit(dab_data)
```

```python
X = dataset_preprocessor.preprocess(dab_data)
Y = dataset_preprocessor.preprocess_target(dab_data)
```

```python
model = KerasAutoencoder.from_dataset_analysis(dataset_preprocessor.get_analysis(), 64, [128, 64])
```

```python
model.autoencoder.summary()
```

```python
model.fit(X, Y, epochs=10, batch_size=32)
```
